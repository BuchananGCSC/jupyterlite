## 1. Setup and the Science <a class="anchor" id="setup"></a>

### The Mars Reconnaissance Orbiter

The [Mars Reconnaissance Orbiter (MRO)](https://mars.nasa.gov/mro/) is a NASA spacecraft that orbits Mars, taking detailed images and gathering data. Like most spacecraft, it is powered by **solar panels**.

In this notebook we'll calculate how much power those solar panels generate — and how that changes depending on which planet the spacecraft is near.

### Key terms

| Term | Definition | Unit |
|------|-----------|------|
| **Power** | Energy transferred per second | Watts (W) |
| **Luminosity** | Total power output of a star | Watts (W) |
| **Brightness** | Power received per square metre at a given distance | W/m² |

### The physics

The Sun radiates energy outward in all directions. As you get farther away, that energy is spread over a larger and larger sphere — so the **brightness** (energy per m²) drops off. This is called the **inverse square law**:

$$b = \frac{L}{4\pi r^2}$$

where $b$ is brightness, $L$ is the Sun's luminosity, and $r$ is the distance from the Sun.

If you **double** the distance, brightness drops by a factor of $2^2 = 4$. If you **triple** the distance, it drops by $3^2 = 9$. That's why Mars gets so much less sunlight than Earth!

▶ **Run this setup cell first:**

````python
import numpy as np

# Physical constants
sigma = 5.670373e-8   # Stefan-Boltzmann constant (Wm^-2K^-4)
Rsun  = 696340000     # Radius of the Sun (m)
Tsun  = 5778          # Surface temperature of the Sun (K)
AU    = 1.495978707e11  # 1 Astronomical Unit in metres (Earth-Sun distance)

# MRO solar panel specs
panel_area       = 10   # combined panel area (m^2)
panel_efficiency = 0.3  # 30% efficient at converting sunlight to electricity
bulb_watts       = 60   # watts in a standard lightbulb (for comparison)

print("Setup complete!")
````

---
## 2. Calculating Brightness <a class="anchor" id="brightness"></a>

### Step 1: Calculate the Sun's luminosity

We can calculate how much power the Sun outputs using the **Stefan-Boltzmann Law**:

$$L = 4\pi R^2 \sigma T^4$$

where $R$ is the radius of the Sun and $T$ is its surface temperature.

````python
# Calculate the Sun's luminosity
# Surface area of a sphere = 4 * pi * r^2
Lsun = 4 * np.pi * Rsun**2 * sigma * Tsun**4

print(f"Sun's luminosity: {Lsun:.3e} W")
# You should get about 3.85 x 10^26 W — an enormous amount of power!
````

### Step 2: Calculate brightness at each planet

Now we apply the inverse square law to find the brightness at each planet's orbit.

The distances below are in **Astronomical Units (AU)** — we need to convert to metres by multiplying by `AU`.

````python
# Planet names and their distances from the Sun in AU
# Note: Neptune and Uranus are deliberately swapped here — we'll deal with that later!
names     = np.array(['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Neptune', 'Uranus', 'Pluto'])
distances = np.array([0.4, 0.7, 1.0, 1.5, 5.2, 9.6, 30, 19.2, 39.5])  # AU

# Calculate brightness at each planet using the inverse square law: b = L / (4 * pi * r^2)
# We multiply distances by AU to convert from AU to metres
brightness = Lsun / (4 * np.pi * (distances * AU)**2)   # W/m^2

print("Brightness at each planet (W/m^2):")
print(brightness)
````

### Step 3: Calculate the MRO solar panel power

Power from the panels = brightness × panel area × panel efficiency

````python
# Calculate power generated by the solar panels at each planet
power = brightness * panel_area * panel_efficiency   # Watts

# How many 60W lightbulbs could this power?
bulb_ratio = power / bulb_watts

print("Power from MRO solar panels (W):")
print(power)
print("\nEquivalent number of 60W lightbulbs:")
print(bulb_ratio)
````

## 3. Looping with enumerate <a class="anchor" id="enumerate"></a>

The raw numbers above are hard to read. Let's print a nice sentence for each planet.

We need to loop over `names` AND use each planet's index to look up the right value in the `power` array. This is exactly what `enumerate` is for!

```python
for i, item in enumerate(my_list):
    print(i)     # the index (0, 1, 2, ...)
    print(item)  # the value at that index
```

▶ **Run this example first to see how enumerate works:**

````python
# enumerate example
fruits = ['apple', 'banana', 'cherry']

for i, fruit in enumerate(fruits):
    print(f"Index {i}: {fruit}")
````

````python
# Now use enumerate to print a sentence for each planet
# i gives us the index so we can look up power[i] and bulb_ratio[i]

for i, name in enumerate(names):
    print(f"At {name} the power is {power[i]} W which is equivalent to {bulb_ratio[i]} lightbulbs!")
````

That works, but the numbers have too many decimal places to read easily. Let's fix that next.

---
## 4. Formatting Numbers Neatly <a class="anchor" id="formatting"></a>

Inside an f-string `{}`, you can add a **format code** after a colon `:` to control how numbers are displayed.

| Format code | What it does | Example |
|-------------|-------------|--------|
| `:.0f` | Float, 0 decimal places | `1826.3` → `1826` |
| `:.2f` | Float, 2 decimal places | `8.3456` → `8.35` |
| `:.3e` | Scientific notation, 3 decimal places | `385000` → `3.850e+05` |
| `:<10` | Left-align text in a 10-character space | `"Mars"` → `"Mars      "` |

▶ **Run this to see formatting in action:**

````python
big_number = 25675.263671
small_number = 0.04388

print(f"No formatting:       {big_number}")
print(f"0 decimal places:    {big_number:.0f}")
print(f"2 decimal places:    {big_number:.2f}")
print(f"Scientific notation: {big_number:.3e}")
print()
print(f"Small number (:.1f): {small_number:.1f}")   # rounds to 0.0 — not ideal!
print(f"Small number (:.4f): {small_number:.4f}")   # better for small values
````

````python
# Now let's improve our planet loop with better formatting
# {name:<8} left-aligns the name in an 8-character space — keeps columns tidy
# {power[i]:.0f} rounds the power to a whole number
# {bulb_ratio[i]:.1f} rounds the bulb ratio to 1 decimal place

for i, name in enumerate(names):
    print(f"At {name:<8} the power is {power[i]:.0f} W which is equivalent to {bulb_ratio[i]:.1f} lightbulbs!")
````

Much more readable! Notice that Pluto shows `0.0 lightbulbs` — the number is so small that `:.1f` rounds it to zero. For very small numbers, `:.3f` or scientific notation works better.

## 5. Building a Function <a class="anchor" id="function"></a>

So far we've been doing all the calculations for every planet at once. But what if we just want to check one specific planet? Let's wrap everything into a **function**.

We'll use a **dictionary** to store the planet distances (as we learned in the Light Speed notebook), so the user just has to provide the planet name.

````python
# Dictionary of planet distances from the Sun in AU
planet_distances = {
    'Mercury': 0.4,
    'Venus':   0.7,
    'Earth':   1.0,
    'Mars':    1.5,
    'Jupiter': 5.2,
    'Saturn':  9.6,
    'Uranus':  19.2,
    'Neptune': 30.0,
    'Pluto':   39.5
}

def solar_panel_power(name):
    """Calculate and print the MRO solar panel power at a given planet.
    
    Args:
        name -- string, must be a key in planet_distances
    """
    # Look up the distance from the dictionary
    distance = planet_distances[name]   # distance in AU
    
    # Calculate Sun's luminosity
    Lsun = 4 * np.pi * Rsun**2 * sigma * Tsun**4
    
    # Calculate brightness using the inverse square law
    b = Lsun / (4 * np.pi * (distance * AU)**2)   # W/m^2
    
    # Calculate panel power and lightbulb ratio
    panel_power = b * panel_area * panel_efficiency
    ratio = panel_power / bulb_watts
    
    print(f"At {name:<8} the power is {panel_power:.0f} W which is equivalent to {ratio:.1f} lightbulbs!")


# Test it
solar_panel_power('Mars')
solar_panel_power('Earth')
solar_panel_power('Jupiter')
````

## 6. Default Argument Values <a class="anchor" id="defaults"></a>

Right now our function always uses the MRO's panel area (10 m²) and efficiency (0.3). What if we wanted to test a different spacecraft with different panels?

We can add **optional arguments with default values**. If the caller doesn't provide them, the defaults are used automatically:

```python
def my_function(required_arg, optional_arg=default_value):
    ...
```

**Example:**
```python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Mars")              # uses default: Hello, Mars!
greet("Mars", "Greetings") # overrides default: Greetings, Mars!
```

▶ **Let's upgrade our function with optional panel settings:**

````python
def solar_panel_power_v2(name, area=10, efficiency=0.3):
    """Calculate and print the solar panel power at a given planet.
    
    Args:
        name       -- string, must be a key in planet_distances
        area       -- float, combined panel area in m^2 (default: 10)
        efficiency -- float, panel efficiency 0-1 (default: 0.3)
    """
    distance = planet_distances[name]
    Lsun = 4 * np.pi * Rsun**2 * sigma * Tsun**4
    b = Lsun / (4 * np.pi * (distance * AU)**2)
    panel_power = b * area * efficiency
    ratio = panel_power / bulb_watts
    
    print(f"At {name:<8} | area={area}m² | eff={efficiency} | power={panel_power:.0f} W ({ratio:.1f} bulbs)")


# Called with no optional args — uses defaults (MRO specs)
print("--- MRO (default panels) ---")
solar_panel_power_v2('Mars')

print()

# Called with custom panel specs — imagine a bigger, more efficient spacecraft
print("--- Hypothetical larger spacecraft ---")
solar_panel_power_v2('Mars', area=25, efficiency=0.45)
````
