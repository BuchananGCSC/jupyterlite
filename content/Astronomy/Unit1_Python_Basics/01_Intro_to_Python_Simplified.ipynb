## 4. Maths <a class="anchor" id="maths"></a>

Python can do maths! Here are the basic operators:

| Symbol | Operation | Example |
|--------|-----------|--------|
| `+` | Addition | `2 + 3` = 5 |
| `-` | Subtraction | `5 - 2` = 3 |
| `*` | Multiplication | `4 * 3` = 12 |
| `/` | Division | `10 / 2` = 5.0 |
| `**` | Power (exponent) | `2 ** 3` = 8 |

▶ **Run this cell:**

````python
# Basic maths — results are stored in new variables
a = 10 + 5   # addition
b = 10 - 5   # subtraction
c = 10 * 5   # multiplication
d = 10 / 5   # division
e = 10 ** 2  # 10 squared (10 to the power of 2)

print(f"10 + 5 = {a}")
print(f"10 - 5 = {b}")
print(f"10 x 5 = {c}")
print(f"10 / 5 = {d}")
print(f"10 squared = {e}")
````

````python
# Let's do some real astronomy maths!
# The speed of light is 299,792,458 metres per second
# The distance from the Earth to the Moon is 384,400 km = 384,400,000 metres

speed_of_light = 299792458   # metres per second
earth_to_moon = 384400000    # metres

# How long does light take to reach the Moon? (time = distance / speed)
time_seconds = earth_to_moon / speed_of_light

print(f"Light takes {time_seconds:.2f} seconds to reach the Moon")
# Note: :.2f means 'show 2 decimal places'
````

## 5. Lists <a class="anchor" id="lists"></a>

A **list** stores multiple values in one variable. Lists use square brackets `[]` and commas between items.

```python
my_list = [item1, item2, item3]
```

**Indexing:** Each item has a position number called an **index**. 
⚠️ Python starts counting from **0**, not 1!

```
planets = ["Mercury", "Venus", "Earth", "Mars"]
index:         0          1        2       3
```

````python
# A list of planet names
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]

print(planets)           # Print the whole list
print(planets[0])        # First item (index 0)
print(planets[3])        # Fourth item (index 3)
print(planets[-1])       # Last item (negative index counts from the end!)
````

````python
# You can add items to a list with .append()
planets.append("Pluto")  # Pluto wants back in!
print(planets)
print(f"Now there are {len(planets)} items in the list")
````

## 6. For Loops <a class="anchor" id="loops"></a>

A **for loop** repeats code for each item in a list. This saves you writing the same thing over and over!

```python
for item in my_list:
    # do something with item
    print(item)
```

⚠️ The code inside the loop **must be indented** (tabbed in). Python uses indentation to know what's inside the loop.

````python
# Print every planet using a for loop
for planet in planets:
    print(planet)   # This line is indented — it runs once for each planet
````

````python
# You can also loop over numbers using range()
# range(5) gives you the numbers 0, 1, 2, 3, 4

for i in range(5):
    print(f"Count: {i}")
````

## 7. NumPy Arrays <a class="anchor" id="numpy"></a>

**NumPy** is a library (a collection of extra tools) that makes working with numbers much easier.

A **NumPy array** is like a list, but you can do maths on the whole array at once — very useful for science!

First, we need to **import** NumPy:

````python
# Import NumPy - we give it the nickname 'np' to save typing
import numpy as np

print("NumPy imported successfully!")
````

````python
# Create a NumPy array of numbers 1 to 8
# np.arange(start, stop) creates numbers from start up to (but not including) stop
planet_numbers = np.arange(1, 9)
print("Planet numbers:", planet_numbers)
````

````python
# The power of arrays: do maths on ALL items at once!
# With a regular list, you'd need a for loop. With NumPy, it's one line:

doubled = planet_numbers * 2
squared = planet_numbers ** 2

print("Original:", planet_numbers)
print("Doubled: ", doubled)
print("Squared: ", squared)
````

````python
# NumPy also has useful maths functions
# np.sqrt() calculates the square root
# np.pi gives you the value of pi (π)

print(f"Pi = {np.pi:.4f}")
print(f"Square root of 144 = {np.sqrt(144)}")

# Real astronomy example:
# The distances of the planets from the Sun in Astronomical Units (AU)
# 1 AU = distance from Earth to Sun
distances_au = np.array([0.39, 0.72, 1.0, 1.52, 5.2, 9.58, 19.2, 30.05])

print("\nDistances from Sun (AU):")
print(distances_au)
````
