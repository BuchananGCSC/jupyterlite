## 1. Setup <a class="anchor" id="setup"></a>

Let's start by importing NumPy and setting up our constants. **Run this cell first** — the rest of the notebook depends on it.

````python
import numpy as np

# These are our constants — values we'll use throughout the notebook
c = 299792458          # speed of light in metres per second
sun_earth = 150e9      # average Earth-Sun distance in metres (1.5 x 10^11)
moon_earth = 384400e3  # average Earth-Moon distance in metres

print("Setup complete!")
print(f"Speed of light: {c:,} m/s")
print(f"Earth-Sun distance: {sun_earth:.2e} m")
print(f"Earth-Moon distance: {moon_earth:.2e} m")
````

## 2. Functions <a class="anchor" id="functions"></a>

Imagine you want to calculate the travel time for the Sun, the Moon, and several other objects. You'd end up writing the same calculation over and over.

A **function** solves this — you write the code once, give it a name, and then call it whenever you need it.

### How to define a function

```python
def function_name(input1, input2):
    # code goes here, indented
    result = input1 + input2
    return result   # sends the result back to whoever called the function
```

- `def` tells Python you're defining a function
- The **inputs** (called *arguments* or *parameters*) go inside the brackets
- `return` sends a value back out of the function
- Everything inside must be **indented**

### How to call a function

```python
answer = function_name(5, 3)   # runs the function with input1=5, input2=3
print(answer)                  # prints 8
```

▶ **Let's write our first function — it calculates travel time for light:**

````python
def light_travel_time(name, distance):
    # This function takes the name of an object and its distance from Earth
    # It calculates the time for light to travel that distance
    # and prints a sentence with the result

    time_seconds = distance / c       # time = distance / speed
    time_minutes = time_seconds / 60  # convert to minutes

    print(f"It takes light {time_minutes:.2f} minutes to reach Earth from the {name}")

# Now let's call the function!
light_travel_time("Sun", sun_earth)
````

````python
# Call the function again with the Moon's distance
light_travel_time("Moon", moon_earth)
````

The Moon's result shows `0.02 minutes` — that's a very small number. It would make more sense to display it in **seconds** instead. We'll fix this in the next section using if/else.

## 3. If / Else Statements <a class="anchor" id="ifelse"></a>

An **if/else statement** lets your code make a decision:

```python
if condition:
    # runs if condition is True
else:
    # runs if condition is False
```

You can also chain multiple conditions with `elif` (short for "else if"):

```python
if condition_1:
    # runs if condition_1 is True
elif condition_2:
    # runs if condition_1 is False BUT condition_2 is True
else:
    # runs if none of the above are True
```

**Common comparison operators:**
| Symbol | Meaning |
|--------|---------|
| `<` | less than |
| `>` | greater than |
| `==` | equal to |
| `<=` | less than or equal to |

▶ **Run this simple example:**

````python
temperature = -60  # degrees Celsius (Mars average!)

if temperature < 0:
    print("It's freezing cold!")
elif temperature < 20:
    print("It's a bit chilly.")
else:
    print("It's warm!")
````

## 4. Improving our Function <a class="anchor" id="improve"></a>

Now let's combine functions and if/else to fix our earlier problem.

We want the function to automatically choose the best unit:
- If the travel time is **less than 60 seconds** → display in **seconds**
- If the travel time is **more than half a year** (~15,778,800 seconds) → display in **years**
- Otherwise → display in **minutes**

▶ **Here's the improved function:**

````python
def light_travel_time_v2(name, distance):
    # Calculates light travel time and picks the best unit automatically

    time_seconds = distance / c  # always calculate in seconds first

    # Now decide which unit to display
    if time_seconds < 60:
        # Less than a minute — use seconds
        time_display = time_seconds
        unit = "seconds"

    elif time_seconds > 15778800:
        # More than half a year — use years
        time_display = time_seconds / 60 / 60 / 24 / 365.25
        unit = "years"

    else:
        # Otherwise — use minutes
        time_display = time_seconds / 60
        unit = "minutes"

    print(f"It takes light {time_display:.2f} {unit} to reach Earth from the {name}")


# Test all three cases:
light_travel_time_v2("Moon", moon_earth)          # should be seconds
light_travel_time_v2("Sun", sun_earth)            # should be minutes
light_travel_time_v2("Alpha Centauri", 4.068e16)  # should be years
````

Notice how the same function now handles all three cases cleanly. The `if/elif/else` block does the decision-making for us.

You should get roughly **1.28 seconds**, **8.34 minutes**, and **4.30 years**.

## 5. Dictionaries <a class="anchor" id="dicts"></a>

A **dictionary** stores pairs of data: a **key** and a **value**.

Think of it like a real dictionary — you look up a word (the key) and find its definition (the value).

```python
my_dict = {
    "key1": value1,
    "key2": value2,
    "key3": value3
}
```

To look up a value, use the key in square brackets:
```python
my_dict["key1"]   # returns value1
```

Dictionaries are perfect for storing distances — the planet name is the key, the distance is the value:

````python
# A dictionary of distances from Earth (in metres)
distances = {
    "Moon":           384400e3,
    "Sun":            150e9,
    "Mars":           225e9,      # average distance
    "Jupiter":        779e9,
    "Alpha Centauri": 4.068e16
}

# Look up a value using its key
print(f"Distance to Moon: {distances['Moon']:.2e} metres")
print(f"Distance to Jupiter: {distances['Jupiter']:.2e} metres")
````

````python
# Now we can write a simpler function — the user just provides the NAME
# and the function looks up the distance in the dictionary

def light_travel_time_v3(name):
    # Looks up the distance from the dictionary, then calculates travel time

    distance = distances[name]   # look up the distance using the name as a key
    time_seconds = distance / c

    if time_seconds < 60:
        time_display = time_seconds
        unit = "seconds"
    elif time_seconds > 15778800:
        time_display = time_seconds / 60 / 60 / 24 / 365.25
        unit = "years"
    else:
        time_display = time_seconds / 60
        unit = "minutes"

    print(f"It takes light {time_display:.2f} {unit} to reach Earth from the {name}")


# Much simpler to call now — just the name!
light_travel_time_v3("Moon")
light_travel_time_v3("Sun")
light_travel_time_v3("Alpha Centauri")
````

## 6. List Comprehension <a class="anchor" id="listcomp"></a>

What if we want to run our function on **every object** in the dictionary at once?

We could write a for loop:
```python
for name in distances:
    light_travel_time_v3(name)
```

Python also has a shorthand for this called **list comprehension** — it's a compact way to apply something to every item in a list, written on a single line:

```python
[do_something(item) for item in my_list]
```

▶ **Compare these two approaches — they do the same thing:**

````python
# Approach 1: for loop (more familiar)
print("--- For loop ---")
for name in distances:
    light_travel_time_v3(name)
````

````python
# Approach 2: list comprehension (more compact)
print("--- List comprehension ---")
[light_travel_time_v3(name) for name in distances]
````

Both produce the same output. List comprehension is especially useful when you want to **collect the results** in a new list:

```python
doubled = [x * 2 for x in [1, 2, 3, 4, 5]]
# doubled = [2, 4, 6, 8, 10]
```

````python
# A quick list comprehension example with numbers
numbers = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in numbers]
squared = [x ** 2 for x in numbers]

print("Original:", numbers)
print("Doubled: ", doubled)
print("Squared: ", squared)
````
