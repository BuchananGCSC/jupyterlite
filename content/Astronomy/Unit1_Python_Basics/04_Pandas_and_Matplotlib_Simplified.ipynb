## 1. Introducing Pandas <a class="anchor" id="pandas"></a>

**Pandas** is a Python library for working with data stored in tables — like a spreadsheet, but inside Python.

The main object in Pandas is called a **DataFrame** — think of it as a table with rows and columns. Each column is a **Series** (similar to a NumPy array).

Why use Pandas instead of just a list?
- You can select rows and columns by **name** (not just index number)
- It handles **missing data** (shown as `NaN`) gracefully
- It can load data directly from CSV files in one line

▶ **Import Pandas and NumPy:**

````python
import pandas as pd    # pd is the standard nickname for Pandas
import numpy as np
import matplotlib.pyplot as plt   # we'll use this for plotting later

print("Libraries imported successfully!")
````

## 2. Loading and Exploring Data <a class="anchor" id="loading"></a>

We have a file called **planets.csv** containing NASA data on the planets and some moons. A CSV (comma-separated values) file is a simple text file where each row is a data entry and each value is separated by a comma.

Before running the code below, take a moment to open **planets.csv** in your file browser and have a look at its structure.

▶ **Load the CSV file into a Pandas DataFrame:**

````python
# pd.read_csv() loads a CSV file and returns a DataFrame
# We give it the filename as a string
df = pd.read_csv('planets.csv')

print("File loaded! Shape:", df.shape)   # (rows, columns)
print("That's", df.shape[0], "rows and", df.shape[1], "columns")
````

````python
# See all the column names
print("Column names:")
print(df.columns.tolist())
# More info on what each column means is in planets.txt
````

````python
# df.describe() gives a quick statistical summary of all numeric columns
# (count, mean, min, max, etc.)
df.describe()
````

You'll notice some cells show **NaN** — this stands for "Not a Number" and means the data is missing for that entry. Pandas handles NaN automatically in most calculations, which is one reason it's so useful.

## 3. Selecting and Modifying Data <a class="anchor" id="selecting"></a>

### Selecting a column

Use the column name in square brackets — just like a dictionary key:

```python
df['column_name']
```

````python
# Select the 'mass' column
masses = df['mass']
print("Masses (10^24 kg):")
print(masses)
````

````python
# You can do maths on an entire column at once — just like a NumPy array!
# The diameter column is in km. Let's calculate the radius (diameter / 2)
radii = df['diameter'] / 2
print("Radii (km):")
print(radii)
````

````python
# We can add this as a new column to the dataframe
df['radius'] = df['diameter'] / 2

# Check it was added
df.head()
````

### Selecting rows

Use `df.loc[]` to filter rows based on a condition:

```python
df.loc[df['column_name'] == 'value']
```

This reads as: *"locate all rows where the column equals this value"*.

````python
# Select only the row for Earth
earth = df.loc[df['planet'] == 'Earth']
earth
````

````python
# The 'moon' column contains 'y' for moons and 'n' for planets
# Select only the moons
moons = df.loc[df['moon'] == 'y']
print(f"Found {len(moons)} moons:")
moons
````

````python
# Select only the planets (moon == 'n')
planets = df.loc[df['moon'] == 'n']
print(f"Found {len(planets)} planets/dwarf planets:")
planets
````

## 4. Introducing Matplotlib <a class="anchor" id="matplotlib"></a>

**Matplotlib** is Python's most popular plotting library. We use the `pyplot` module (imported as `plt`) to create charts quickly.

The basic pattern for any plot is:
1. Call a plot function (e.g. `plt.scatter()`, `plt.plot()`)
2. Add labels and a title
3. Show or save the figure

**Common plot types:**
| Function | Chart type |
|----------|------------|
| `plt.scatter(x, y)` | Scatter plot |
| `plt.plot(x, y)` | Line chart |
| `plt.bar(x, height)` | Bar chart |
| `plt.hist(data)` | Histogram |

▶ **A simple scatter plot:**

````python
# Let's plot mass vs radius for all the moons
# First extract the columns we need
moon_radii = moons['radius']
moon_masses = moons['mass']

# Create a scatter plot — x axis first, then y axis
plt.scatter(moon_radii, moon_masses)

# Add axis labels and a title
plt.xlabel('Radius (km)')
plt.ylabel('Mass (10^24 kg)')
plt.title('Mass vs Radius for Solar System Moons')

# Display the plot
plt.show()
````

We have a plot! But we can't tell which dot is which moon. Let's learn how to customise it.

---
## 5. Building a Complete Plot <a class="anchor" id="complete"></a>

### Customising scatter plots

`plt.scatter()` accepts many optional arguments to change the appearance:

| Argument | What it does | Example |
|----------|-------------|--------|
| `color` | Sets the dot colour | `color='red'` |
| `s` | Sets the dot size | `s=100` |
| `alpha` | Sets transparency (0=invisible, 1=solid) | `alpha=0.7` |
| `label` | Legend label for this data series | `label='Moons'` |

### Adding labels to individual points

`plt.text(x, y, "label")` adds a text annotation at position (x, y).

````python
# A more complete plot with labels on each moon

# Extract the data we need
moon_names  = moons['planet'].values   # .values converts the Pandas Series to a plain array
moon_radii  = moons['radius'].values
moon_masses = moons['mass'].values

# Create the scatter plot with some styling
plt.scatter(moon_radii, moon_masses, color='steelblue', s=80, alpha=0.8)

# Add a name label next to each point
for i, name in enumerate(moon_names):
    plt.text(moon_radii[i] + 20, moon_masses[i], name, fontsize=9)
    # The + 20 shifts the label slightly to the right of the dot

# Labels and title
plt.xlabel('Radius (km)')
plt.ylabel('Mass (10^24 kg)')
plt.title('Mass vs Radius for Solar System Moons')

plt.show()
````

### Plotting multiple datasets on one chart

You can call `plt.scatter()` multiple times before `plt.show()` to overlay datasets. Use `label=` and `plt.legend()` to tell them apart.

````python
# Plot planets and moons on the same chart
plt.scatter(planet_radii, planet_masses, color='royalblue', s=100, label='Planets')
plt.scatter(moon_radii,   moon_masses,   color='tomato',    s=60,  label='Moons', alpha=0.8)

plt.xlabel('Radius (km)')
plt.ylabel('Mass (10^24 kg)')
plt.title('Mass vs Radius: Planets and Moons')
plt.legend()   # adds a legend using the label= values

plt.show()
````

Jupiter dominates the chart! Its mass is so much larger than everything else. A **log scale** can help when data spans a huge range.

````python
# The same plot but with a logarithmic y-axis
plt.scatter(planet_radii, planet_masses, color='royalblue', s=100, label='Planets')
plt.scatter(moon_radii,   moon_masses,   color='tomato',    s=60,  label='Moons', alpha=0.8)

plt.xlabel('Radius (km)')
plt.ylabel('Mass (10^24 kg) — log scale')
plt.title('Mass vs Radius: Planets and Moons (log scale)')
plt.yscale('log')   # switch y-axis to log scale
plt.legend()

plt.show()
````

## 6. Saving Your Work <a class="anchor" id="saving"></a>

### Saving a plot

Use `plt.savefig('filename.png')` **before** `plt.show()` to save the figure to a file.

````python
# Build a plot and save it
plt.scatter(moon_radii, moon_masses, color='steelblue', s=80)

for i, name in enumerate(moon_names):
    plt.text(moon_radii[i] + 20, moon_masses[i], name, fontsize=9)

plt.xlabel('Radius (km)')
plt.ylabel('Mass (10^24 kg)')
plt.title('Mass vs Radius for Solar System Moons')

plt.savefig('moon_plot.png')   # saves the file — must come before plt.show()
plt.show()

print("Plot saved as moon_plot.png")
````
